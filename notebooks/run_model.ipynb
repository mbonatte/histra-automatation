# Run

```python
from histra_automation.processing_steps import prepare_model
from histra_automation.run_scenario import run_scenario

from concurrent.futures import ThreadPoolExecutor, as_completed
def run_model(input_path, scenarios, mode='local', timeout=360, max_workers=4, **kwargs):
    """Run all scenarios in parallel."""
    prepare_model(input_path)

    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = []
        for i, scenario in enumerate(scenarios):
            time.sleep(0.2)
            futures.append(executor.submit(run_scenario, input_path, scenario, i, mode, timeout, **kwargs))
        for future in as_completed(futures):
            try:
                future.result()
            except KeyboardInterrupt:
                print("\n🛑 Program interrupted by user. Cancelling all remaining tasks...")
                executor.shutdown(wait=False, cancel_futures=True)
                raise
            except Exception as e:
                print(f"Error in one of the scenarios: {e}")
                time.sleep(0.2)
```

```python
import tempfile
import shutil

MODEL_NAME = "Bridge_2"
INPUT_FILE = r"Bridge_2\Bridge_2.hrx"


INPUT_FILE = os.path.abspath(INPUT_FILE)

directory = os.path.dirname(INPUT_FILE)

with tempfile.TemporaryDirectory(prefix="histra_run_", dir=directory) as temp_dir:
    print(temp_dir)
    input_copy = os.path.join(temp_dir, os.path.basename(INPUT_FILE))
    shutil.copy2(INPUT_FILE, input_copy)
    run_model(input_copy, scenarios, mode='local'
              , timeout=30, max_workers=2, all_steps=True)
```
